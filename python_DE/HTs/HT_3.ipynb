Пограничные случаи: \

```python
df['add-ons_purchased'] = df['add-ons_purchased'].fillna(0)
```

```python
payment_df = (
    df.groupby(['customer_id', 'payment_method'])
    .size()
    .reset_index(name='count')
)

payment_df = payment_df.sort_values(by=['count', 'payment_method'], ascending=[False, True])
payment_df
```

```python
# Сначала находим наиболее частый preferred_payment для каждого customer_id
most_frequent = payment_df.groupby('customer_id')['payment_method'].agg(
    most_popular=pd.Series.mode).reset_index()

# Теперь оставляем только эти строки и объединяем с исходным DataFrame
preferred_payment = payment_df.merge(most_frequent, on='customer_id', how='left')
```

```python
customer_stats = (
    df.groupby('customer_id')
    .agg(
        total_spent=('total_price', 'sum'),
        addons_spent=('add-on_total', 'sum'),
        orders_count=('total_price', 'count')
    )
    .reset_index()
)

result = pd.merge(customer_stats, preferred_payment, on='customer_id')
result = result[['customer_id', 'payment_method', 'total_spent', 'addons_spent', 'orders_count']]

top_10_customers = result.sort_values(by='total_spent', ascending=False).head(10).reset_index(drop=True)

print(top_10_customers)
```

## 2.
Постройте пайплайн на Pandas (без циклов по строкам):
 
 1. Очистка: типы дат, пропуски, аномалии (отрицательные суммы/нулевые цены)
 2. Метрики:\
    • доход по методу доставки;\
    • доход по типу продукта;\
    • доход по доп. услугам: по месяцам и по кварталам;\
    • cohort-анализ: когорта = месяц первого заказа покупателя; для каждой когорты - выручка и число покупателей в следующие 0, 1, 2, 3 месяца.\
Дополнительно: оформите шаги 1–2 как функции `clean(df) → df` и `build_metrics(df) → dict[str, DataFrame]`.
 3. Визуализация (минимум 3 графика): bar (доставка / тип продукта), line (доп. услуги по месяцам), heatmap когортной матрицы выручки.
 4. Выводы: несколько предложений по графикам.

```python
import numpy as np
```

```python
def clean(df: pd.DataFrame) -> pd.DataFrame:
    """ Очистка данных. Возвращает очищенный DataFrame. """
    df_new = df.copy()

    date_cols = [column for column in df_new.columns if 'date' in column.lower()]
    for col in date_cols:
        if col in df_new.columns:
            df_new[col] = pd.to_datetime(df_new[col], errors='coerce')

    # пропуски и аномалии
    if ['add-ons_purchased'] in df_new.columns:
        df_new[col] = df_new[col].fillna(0)

    if 'total_price' in df_new.columns:
        df_new = df_new[df_new['total_price'] > 0]

    if 'add-on_total' in df_new.columns:
        df_new = df_new[df_new['add-on_total'] >= 0]

    return df_new


def build_metrics(df: pd.DataFrame) -> dict[str, pd.DataFrame]:
    """
    Строит метрики по клиентам:
      - preferred_payment: самый частый метод оплаты; при ничьей — лексикографически первый.
      - общая статистика: total_spent, addons_spent, orders_count.
    
    Возвращает dict:
      {
        "preferred_payment": DataFrame(customer_id, payment_method),
        "customer_stats": DataFrame(customer_id, total_spent, addons_spent, orders_count),
        "top_10": DataFrame(топ-10 клиентов по total_spent)
      }
    """
    customer_stats = (
        df.groupby('customer_id', as_index=False)
        .agg(
            total_spent=('total_price', 'sum'),
            addons_spent=('add-on_total', 'sum'),
            orders_count=('total_price', 'count')
        )
    )

    payment_counts = (
        df.groupby(['customer_id', 'payment_method'], as_index=False)
        .size()
        .rename(columns={'size': 'count'})
    )

    preferred_payment = (
        payment_counts.sort_values(['count', 'payment_method'], ascending=[False, True])
        .groupby('customer_id')
        .head(1)[['customer_id', 'payment_method']]
        .reset_index(drop=True)
    )

    result = customer_stats.merge(preferred_payment, on='customer_id', how='left')

    top_10 = (
        result.sort_values('total_spent', ascending=False)
        .head(10)
        .reset_index(drop=True)
    )

    return {
        "preferred_payment": preferred_payment,
        "customer_stats": customer_stats,
        "top_10": top_10
    }
```
